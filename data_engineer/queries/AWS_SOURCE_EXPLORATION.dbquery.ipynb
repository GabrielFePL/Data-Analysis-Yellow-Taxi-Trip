```sql
SHOW SCHEMAS IN aws_glue_nyc_taxi;

SHOW TABLES IN aws_glue_nyc_taxi.nyc_taxi_study;

DESCRIBE TABLE EXTENDED
  aws_glue_nyc_taxi.nyc_taxi_study.yellow_tripdata;

SELECT
  year,
  month,
  COUNT(*) AS trip_count,
  MIN(tpep_pickup_datetime) AS min_pickup,
  MAX(tpep_pickup_datetime) AS max_pickup
FROM aws_glue_nyc_taxi.nyc_taxi_study.yellow_tripdata
GROUP BY year, month
ORDER BY year, month;

SELECT
  COUNT(*) AS trip_count,
  COUNT(DISTINCT year, month) AS published_months,
  MIN(tpep_pickup_datetime) AS min_pickup,
  MAX(tpep_pickup_datetime) AS max_pickup,
  SUM(total_amount) AS gross_revenue
FROM aws_glue_nyc_taxi.nyc_taxi_study.yellow_tripdata;
```
